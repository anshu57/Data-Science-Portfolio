```python
bill_2018 = bill_df['2016':'2018'][['Billed_amount']]
```

```python
bill_2018
```

Simple Moving Average = (t+(t-1)+(t-2)+...+(t-n))/n

```python
bill_2018['Billed_amount'].rolling(window=3).mean()
```

```python
bill_2018['ma_rolling_3'] = bill_2018['Billed_amount'].rolling(window=3).mean().shift(1)
```

```python
bill_2018
```

```python
bill_2018.plot()
```

Weighted moving Average = (t*weightage factor) + ((t-1)*weightage factor -1) +((t-n)*weightage factor-n)/n

```python
def wma(weights):
    def calc(x):
        return (weights*x).mean()
    return calc
```

```python
bill_2018['Billed_amount'].rolling(window=3).apply(wma(np.array([0.5,1,1.5])))
```

```python
bill_2018['wma_rolling_3'] = bill_2018['Billed_amount'].rolling(window=3).apply(wma(np.array([0.5,1,1.5]))).shift(1)
```

```python
bill_2018
```

```python
bill_2018.plot()
```

Explonential moving average = (close - previous EMA)*(2/(span+1)) + previous EMA

```python
bill_2018['Billed_amount'].ewm(span=3, adjust=False, min_periods=0).mean()
```

```python
bill_2018['ewm_window_3'] = bill_2018['Billed_amount'].ewm(span=3, adjust=False, min_periods=0).mean().shift(1)
```

```python
bill_2018
```

```python
bill_2018.plot()
```

Explonential Smoothing

```python
bill_2018['Billed_amount'].ewm(alpha=0.7, adjust=False, min_periods=3).mean()
```

```python
bill_2018['ewm_window_3_7'] = bill_2018['Billed_amount'].ewm(alpha=0.7, adjust=False, min_periods=3).mean().shift(1)
```

```python
bill_2018
```

```python
bill_2018[['Billed_amount', 'ewm_window_3_7']].plot()
```

```python
bill_2018.plot()
```

```python
bill_2018['ewm_window_3_3'] = bill_2018['Billed_amount'].ewm(alpha=0.3, adjust=False, min_periods=3).mean().shift(1)
```

```python
bill_2018
```

```python
bill_2018[['Billed_amount', 'ewm_window_3_3']].plot()
```
